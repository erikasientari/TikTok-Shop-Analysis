```python
# Data Import
import pandas as pd

df = pd.read_csv("tiktok_clean.csv")
```

```python
# Understanding Data
df.head()
```

```python
## Changing Text Format
df.columns = (
    df.columns
    .str.lower()
    .str.strip()
    .str.replace(" ", "_")
)
```

```python
# Check Missing Values
df.isnull().sum()

# Check Duplicates
df[df.duplicated()]
```

```python
# Remove duplicates
df = df.drop_duplicates()

# Handle missing values
df = df.fillna({
    "cancel_reason": "unknown",
    "quantity": 0,
    "payment_method": "unknown",
    "order_amount": 0,
    "order_refund_amount": 0,
    "payment_platform_discount": 0
})
```

```python
# Check
df.isnull().sum()

df.duplicated().sum()
```
