```python
import pandas as pd

df = pd.read_csv("tiktok_cleaned.csv")
```

```python
### Gross Revenue (total order_amount)
gross_revenue = df["order_amount"].sum()

gross_revenue

### Net Revenue (order amount total, exclude if 'cancelled'')
df_not_cancel = df[df["order_status"] != "cancelled"]

net_revenue = df_not_cancel["order_amount"].sum()

net_revenue

### Count: Order total
df["order_id"].nunique()

### Count: total items sold
import pandas as pd

df = pd.read_csv("tiktok_cleaned2.csv")

df['quantity'].sum()
```

```python
### Revenue by Month
# Change Time Format 
df["month"] = df["order_date"].dt.strftime("%Y-%B")

# Collect Data
import pandas as pd

df["order_date"] = pd.to_datetime(df["order_date"])
df["year_month"] = df["order_date"].dt.to_period("M").astype(str)

df_clean = df[df["order_status"] != "cancelled"]

df_monthly = df_clean.groupby("year_month").agg(
    net_revenue=("order_amount", "sum"),
    total_orders=("order_id", "nunique")
).reset_index()

df_monthly.to_csv("monthly_net_kpi.csv", index=False, encoding="utf-8-sig")
```

```python
### Performance Based on Product Category
import pandas as pd

#1 clean
df["product_category"] = df["product_category"].str.strip()

#2 code
df_category = df.groupby("product_category").agg(
    total_orders=("order_id", "nunique"),
    total_revenue=("order_amount", "sum"),
    avg_order_value=("order_amount", "mean")
).reset_index()

#3 sort
df_category = df_category.sort_values("total_revenue", ascending=False)

#4 save
df_category.to_csv("category_tableau_ready.csv", index=False, encoding="utf-8-sig")
```

```python
### Performance Based on Showcase
import pandas as pd

#1 clean product name
df["product_name"] = df["product_name"].str.strip()

df_product = df.groupby("product_name").agg(
    total_orders=("order_id", "nunique"),
    total_revenue=("order_amount", "sum"),
    avg_order_value=("order_amount", "mean")
).reset_index()

#2 Sort by Revenue
df_product = df_product.sort_values("total_revenue", ascending=False)

#3 Save
df_product.to_csv("product_tableau_ready.csv", index=False, encoding="utf-8-sig")
```

```python
### Cancel Rate
total_orders = df["order_substatus"].shape[0]
cancel_orders = (df["order_substatus"] == "cancelled").sum()

cancel_percentage = (cancel_orders / total_orders) * 100

cancel_percentage

## Cancel Frequency
import pandas as pd

df_status = df["order_substatus"].value_counts().reset_index()

df_status.columns = ["order_substatus", "frequency"]

df_status.to_csv("order_substatus_summary.csv", index=False, encoding="utf-8-sig")

## Order Cancellation Reasons and Frequency
import pandas as pd

# count frequency cancel reason
df_cancel_reason = (
    df["cancel_reason"]
    .value_counts()
    .reset_index()
)

# columns rename
df_cancel_reason.columns = ["cancel_reason", "frequency"]

# Sort from Highest
df_cancel_reason = df_cancel_reason.sort_values("frequency", ascending=False)

# Save to CSV format
df_cancel_reason.to_csv(
    "cancel_reason_summary.csv",
    index=False,
    encoding="utf-8-sig"
)

## Cancel Amount by Province
import pandas as pd

df_cancel_province = df[df["order_status"] == "cancelled"]["province"].value_counts().reset_index()

df_cancel_province.columns = ["province", "cancel_frequency"]

df_cancel_province.to_csv("cancel_by_province.csv", index=False, encoding="utf-8-sig")

## Cancel Rate by Province
import pandas as pd

#1 total order per province
total_per_province = df.groupby("province")["order_id"].count()

#2 cancel order per province
cancel_per_province = df[df["order_status"] == "cancelled"].groupby("province")["order_id"].count()

#3 join
cancel_rate = (cancel_per_province / total_per_province) * 100

df_cancel_rate = cancel_rate.reset_index()
df_cancel_rate.columns = ["province", "cancel_rate"]

#4 sort
df_cancel_rate = df_cancel_rate.sort_values("cancel_rate", ascending=False)

#5 save
df_cancel_rate.to_csv("cancel_rate_by_province.csv", index=False, encoding="utf-8-sig")
```
